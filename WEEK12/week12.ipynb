```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor
from sklearn.metrics import accuracy_score, mean_squared_error, r2_score
from sklearn.preprocessing import LabelEncoder

titanic_df = pd.read_csv('titanic_cleaned.csv')

X_titanic = titanic_df.drop('Survived', axis=1)
y_titanic = titanic_df['Survived']

X_train_t, X_test_t, y_train_t, y_test_t = train_test_split(X_titanic, y_titanic, test_size=0.2, random_state=42)

titanic_model = RandomForestClassifier(n_estimators=100, random_state=42)
titanic_model.fit(X_train_t, y_train_t)

y_pred_t = titanic_model.predict(X_test_t)
print("=== Titanic Model ===")
print(f"Accuracy: {accuracy_score(y_test_t, y_pred_t):.2f}")


airbnb_df = pd.read_csv('airbnb_cleaned.csv')

for col in airbnb_df.select_dtypes(include='object').columns:
    airbnb_df[col] = LabelEncoder().fit_transform(airbnb_df[col])

X_airbnb = airbnb_df.drop('price', axis=1)
y_airbnb = airbnb_df['price']

X_train_a, X_test_a, y_train_a, y_test_a = train_test_split(X_airbnb, y_airbnb, test_size=0.2, random_state=42)

airbnb_model = RandomForestRegressor(n_estimators=100, random_state=42)
airbnb_model.fit(X_train_a, y_train_a)

y_pred_a = airbnb_model.predict(X_test_a)
print("\n=== New York Airbnb Model ===")
print(f"R² Score : {r2_score(y_test_a, y_pred_a):.2f}")
print(f"RMSE     : {mean_squared_error(y_test_a, y_pred_a, squared=False):.2f}")

print("""
=== Suggestions to Improve Model Performance ===
1. Feature Engineering  - create new meaningful columns from existing ones
2. Hyperparameter Tuning - use GridSearchCV to find the best model settings
3. Handle Outliers      - remove or cap extreme values especially in price
4. More Data            - larger datasets generally produce better models
5. Try Other Algorithms - XGBoost, SVM, or Neural Networks may perform better
""")
```
