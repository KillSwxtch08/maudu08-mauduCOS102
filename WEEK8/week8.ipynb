```python
import tkinter as tk

class Zenith:
    def __init__(self, name):
        self.name = name

    def mutual_services(self):
        return ["Lines of credit", "Investment management and accounts", "Insurance"]

    def unique_services(self):
        return []

class RetailBanking(Zenith):
    def unique_services(self):
        return ["Retirement and education accounts", "Loans and mortgages", "Checking and saving"]

class GlobalBanking(Zenith):
    def unique_services(self):
        return ["Multi-currency management services and products", "Foreign currency accounts",
                "Foreign currency credit cards", "Transborder advisory services", "Liquidity management"]

class CommercialBanking(Zenith):
    def unique_services(self):
        return ["Advisory services"]

employees = {
    "Mary": RetailBanking("Mary"),
    "Agatha": GlobalBanking("Agatha"),
    "Noel": CommercialBanking("Noel")
}

def search():
    name = entry.get().strip().title()
    result_box.config(state="normal")
    result_box.delete("1.0", tk.END)

    if name not in employees:
        result_box.insert(tk.END, f"Employee '{name}' not found.\nTry: Mary, Agatha or Noel.")
    else:
        emp = employees[name]
        result_box.insert(tk.END, f"Employee: {emp.name}\n\n")
        result_box.insert(tk.END, "Mutual Services:\n")
        for s in emp.mutual_services():
            result_box.insert(tk.END, f"  - {s}\n")
        result_box.insert(tk.END, "\nUnique Services:\n")
        for s in emp.unique_services():
            result_box.insert(tk.END, f"  - {s}\n")

    result_box.config(state="disabled")

root = tk.Tk()
root.title("Zenith Bank")
root.geometry("500x400")

tk.Label(root, text="Zenith Bank Nigeria", font=("Helvetica", 16, "bold")).pack(pady=10)
tk.Label(root, text="Enter Employee Name:").pack()
entry = tk.Entry(root, font=("Helvetica", 12), width=25)
entry.pack(pady=5)
tk.Button(root, text="Search", command=search).pack(pady=5)
result_box = tk.Text(root, font=("Courier", 11), height=15, width=55, state="disabled")
result_box.pack(pady=10)

root.mainloop()
```
