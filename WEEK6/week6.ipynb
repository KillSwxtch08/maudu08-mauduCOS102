```python
import tkinter as tk
import random

class Employee:
    def __init__(self, employees, tasks):
        self.employees = employees
        self.tasks = tasks
        self.attendance = []

    def check_employee(self, name):
        return name in self.employees

    def take_attendance(self, name):
        if name not in self.attendance:
            self.attendance.append(name)

    def assign_task(self):
        index = random.randint(0, len(self.tasks) - 1)
        return self.tasks[index]

    def refuse_access(self):
        return "Sorry, you are not recognized as an employee. Access denied."

employees_list = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali", "Andrew Ugwu",
                   "Stella Mankinde", "Jane Akibo", "Ago James", "Michell Taiwo", "Abraham Jones",
                   "Nicole Anide", "Kosi Korso", "Adele Martins", "Emmanuel Ojo", "Ajayi Fatima"]

tasks_list = ["Loading", "Transporting", "Reviewing Orders", "Customer Service", "Delivering Items"]

emp_system = Employee(employees_list, tasks_list)

def handle_submit():
    name = name_entry.get().strip()
    if emp_system.check_employee(name):
        emp_system.take_attendance(name)
        task = emp_system.assign_task()
        result_label.config(text=f"Welcome, {name}! Attendance recorded.\nToday's task: {task}")
    else:
        result_label.config(text=emp_system.refuse_access())

root = tk.Tk()
root.title("Mrs. Jane's Delivery Service - Employee Check-in")

tk.Label(root, text="Enter your name:").grid(row=0, column=0, padx=10, pady=10)
name_entry = tk.Entry(root, width=30)
name_entry.grid(row=0, column=1, padx=10, pady=10)

tk.Button(root, text="Check In", command=handle_submit).grid(row=1, column=0, columnspan=2, pady=5)

result_label = tk.Label(root, text="", wraplength=300, justify="left")
result_label.grid(row=2, column=0, columnspan=2, padx=10, pady=10)

root.mainloop()
```

```python
import tkinter as tk

class DeliveryService:
    def __init__(self):
        self.rates = {
            "PAU": {"high": 2000, "low": 1500},
            "Epe": {"high": 5000, "low": 4000}
        }

    def calculate_charge(self, location, weight):
        rate = self.rates[location]
        if weight >= 10:
            return rate["high"]
        else:
            return rate["low"]

service = DeliveryService()

def handle_submit():
    location = location_var.get()
    try:
        weight = float(weight_entry.get())
    except ValueError:
        result_label.config(text="Please enter a valid weight.")
        return

    charge = service.calculate_charge(location, weight)
    result_label.config(text=f"Delivery charge: N{charge:,}")

root = tk.Tk()
root.title("Delivery Charge Calculator")

tk.Label(root, text="Location:").grid(row=0, column=0, padx=10, pady=10)
location_var = tk.StringVar(value="PAU")
tk.OptionMenu(root, location_var, "PAU", "Epe").grid(row=0, column=1, padx=10, pady=10)

tk.Label(root, text="Package weight (kg):").grid(row=1, column=0, padx=10, pady=10)
weight_entry = tk.Entry(root)
weight_entry.grid(row=1, column=1, padx=10, pady=10)

tk.Button(root, text="Calculate", command=handle_submit).grid(row=2, column=0, columnspan=2, pady=5)

result_label = tk.Label(root, text="", font=("Arial", 12, "bold"))
result_label.grid(row=3, column=0, columnspan=2, padx=10, pady=10)

root.mainloop()
```
