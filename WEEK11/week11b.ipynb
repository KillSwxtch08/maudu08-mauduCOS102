```python
from abc import ABC, abstractmethod

class coup_de_ecriva(ABC):
    @abstractmethod
    def Fan_Page(self):
        pass

class FC_Cirok(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "FC_Cirok"')

class Madiba_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Madiba_FC"')

class Blue_Jay_FC(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "Blue_Jay_FC"')

class TSG_Walker(coup_de_ecriva):
    def Fan_Page(self):
        print('Welcome to "TSG_Walker"')

clubs = {
    "FC_Cirok": FC_Cirok,
    "Madiba_FC": Madiba_FC,
    "Blue_Jay_FC": Blue_Jay_FC,
    "TSG_Walker": TSG_Walker
}

club_name = input("Enter the club you support: ").strip()

if club_name in clubs:
    obj = clubs[club_name]()
    obj.Fan_Page()
else:
    print(f"Club '{club_name}' not found. Available clubs: {', '.join(clubs.keys())}")
```

```python
import tkinter as tk
from abc import ABC, abstractmethod

class External_Vendors(ABC):
    @abstractmethod
    def menu(self):
        pass

class CooperativeCafeteria(External_Vendors):
    def menu(self):
        return [
            ("Jollof Rice and Stew", 200),
            ("White Rice and Stew", 200),
            ("Fried Rice", 200),
            ("Salad", 100),
            ("Plantain", 100)
        ]

class FaithHostelCafeteria(External_Vendors):
    def menu(self):
        return [
            ("Fried Rice", 400),
            ("White Rice and Stew", 400),
            ("Jollof Rice", 400),
            ("Beans", 200),
            ("Chicken", 1000)
        ]

class StudentCentreCafeteria(External_Vendors):
    def menu(self):
        return [
            ("Chicken Fried Rice", 800),
            ("Pomo Sauce", 300),
            ("Spaghetti Jollof", 500),
            ("Amala/Ewedu", 500),
            ("Semo with Eforiro Soup", 500)
        ]

vendors = {
    "Cooperative Hostel": CooperativeCafeteria(),
    "Faith Hostel": FaithHostelCafeteria(),
    "Student Center": StudentCentreCafeteria()
}

def show_menu():
    vendor_name = var.get()
    vendor = vendors[vendor_name]

    for widget in menu_frame.winfo_children():
        widget.destroy()

    tk.Label(menu_frame, text="Main Meal", font=("Helvetica", 11, "bold"), width=25, anchor="w").grid(row=0, column=0, padx=10)
    tk.Label(menu_frame, text="Price (N)", font=("Helvetica", 11, "bold"), width=10, anchor="w").grid(row=0, column=1)

    for i, (meal, price) in enumerate(vendor.menu(), start=1):
        tk.Label(menu_frame, text=meal, font=("Helvetica", 11), width=25, anchor="w").grid(row=i, column=0, padx=10, pady=2)
        tk.Label(menu_frame, text=str(price), font=("Helvetica", 11), width=10, anchor="w").grid(row=i, column=1, pady=2)

root = tk.Tk()
root.title("PAU External Food Vendors")
root.geometry("450x350")

tk.Label(root, text="PAU Food Vendors", font=("Helvetica", 16, "bold")).pack(pady=10)
tk.Label(root, text="Select a vendor:").pack()

var = tk.StringVar(value="Cooperative Hostel")
for vendor in vendors:
    tk.Radiobutton(root, text=vendor, variable=var, value=vendor).pack()

tk.Button(root, text="Show Menu", command=show_menu).pack(pady=8)

menu_frame = tk.Frame(root)
menu_frame.pack()

root.mainloop()
```
