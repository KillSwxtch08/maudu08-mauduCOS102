```python
import pandas as pd

class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

class Pirates(Student):
    def category(self):
        return "The_Pirates"

class Yankees(Student):
    def category(self):
        return "The_Yankees"

class Bulls(Student):
    def category(self):
        return "The_Bulls"

class Restricted(Student):
    def category(self):
        return "Restricted"

    def get_error(self):
        return f"Error: {self.get_name()} (age {self.get_age()}) is above 25. Access denied."


df = pd.read_csv('SIS.csv')

pirates_list = []
yankees_list = []
bulls_list = []

for _, row in df.iterrows():
    name = row['Name']
    age = row['Age']

    if 14 < age < 18:
        s = Pirates(name, age)
        pirates_list.append({'Name': s.get_name(), 'Age': s.get_age()})

    elif 18 < age < 22:
        s = Yankees(name, age)
        yankees_list.append({'Name': s.get_name(), 'Age': s.get_age()})

    elif 22 < age < 25:
        s = Bulls(name, age)
        bulls_list.append({'Name': s.get_name(), 'Age': s.get_age()})

    elif age > 25:
        s = Restricted(name, age)
        print(s.get_error())

if pirates_list:
    pirates_df = pd.DataFrame(pirates_list)
    pirates_df.to_csv('The_Pirates.csv', index=False)
    print("\n--- The Pirates ---")
    print(pirates_df.to_string(index=False))

if yankees_list:
    yankees_df = pd.DataFrame(yankees_list)
    yankees_df.to_csv('The_Yankees.csv', index=False)
    print("\n--- The Yankees ---")
    print(yankees_df.to_string(index=False))

if bulls_list:
    bulls_df = pd.DataFrame(bulls_list)
    bulls_df.to_csv('The_Bulls.csv', index=False)
    print("\n--- The Bulls ---")
    print(bulls_df.to_string(index=False))
```
