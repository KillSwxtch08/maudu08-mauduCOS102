```python
import numpy as np

print("1. Quadratic (Ax^2 + Bx + C = 0)")
print("2. Cubic (Ax^3 + Bx^2 + Cx + D = 0)")
print("3. Quartic (Ax^4 + Bx^3 + Cx^2 + Dx + E = 0)")
choice = input("Choose operation (1/2/3): ")

if choice == "1":
    coeffs = [float(input("Enter A: ")), float(input("Enter B: ")), float(input("Enter C: "))]
elif choice == "2":
    coeffs = [float(input("Enter A: ")), float(input("Enter B: ")), float(input("Enter C: ")), float(input("Enter D: "))]
elif choice == "3":
    coeffs = [float(input("Enter A: ")), float(input("Enter B: ")), float(input("Enter C: ")), float(input("Enter D: ")), float(input("Enter E: "))]
else:
    coeffs = None
    print("Invalid choice")

if coeffs:
    print("Roots:", np.roots(coeffs))
```
