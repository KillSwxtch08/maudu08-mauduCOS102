```python
def simple_interest():
    P = float(input("Enter Principal (P): "))
    R = float(input("Enter annual Rate in %: "))
    T = float(input("Enter Time in years (T): "))
    A = P * (1 + (R / 100) * T)
    print(f"Interest earned: {A - P:.2f}")
    print(f"Total Amount (A): {A:.2f}")

def compound_interest():
    P = float(input("Enter Principal (P): "))
    R = float(input("Enter annual Rate in %: "))
    n = float(input("Enter times compounded per year (n): "))
    t = float(input("Enter Time in years (t): "))
    r = (R / 100) / n
    A = P * (1 + r) ** (n * t)
    print(f"Interest earned: {A - P:.2f}")
    print(f"Total Amount (A): {A:.2f}")

def annuity_plan():
    PMT = float(input("Enter Payment per period (PMT): "))
    R = float(input("Enter annual Rate in %: "))
    n = float(input("Enter compounding periods per year (n): "))
    t = float(input("Enter Time in years (t): "))
    r = (R / 100) / n
    A = PMT * (((1 + r) ** (n * t) - 1) / r)
    print(f"Future Value of Annuity (A): {A:.2f}")

print("1. Simple Interest")
print("2. Compound Interest")
print("3. Annuity Plan")
choice = input("Enter choice (1/2/3): ")

if choice == "1":
    simple_interest()
elif choice == "2":
    compound_interest()
elif choice == "3":
    annuity_plan()
else:
    print("Invalid choice")
```
