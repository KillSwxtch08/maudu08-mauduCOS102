```python
import tkinter as tk
import csv

# Load employee records — "Name" and "Department" below are placeholders;
# I'll swap these for the real column names once I see your CSV.
employees = []
with open("GIG-logistics.csv", newline="") as f:
    reader = csv.DictReader(f)
    for row in reader:
        employees.append(row)

def check_employee():
    name = name_entry.get().strip()
    department = dept_entry.get().strip()

    match = None
    for emp in employees:
        if emp["Name"].lower() == name.lower() and emp["Department"].lower() == department.lower():
            match = emp
            break

    if match:
        colleagues = [e["Name"] for e in employees
                      if e["Department"].lower() == department.lower() and e["Name"] != name]
        result_label.config(text=f"Welcome, {name}! Other members of {department}: {', '.join(colleagues)}")
    else:
        result_label.config(text=f"{name} does not exist as an employee.")

root = tk.Tk()
root.title("GIG Logistics Employee Verification")

tk.Label(root, text="Name:").grid(row=0, column=0)
name_entry = tk.Entry(root)
name_entry.grid(row=0, column=1)

tk.Label(root, text="Department:").grid(row=1, column=0)
dept_entry = tk.Entry(root)
dept_entry.grid(row=1, column=1)

tk.Button(root, text="Verify", command=check_employee).grid(row=2, column=0, columnspan=2)

result_label = tk.Label(root, text="", wraplength=300)
result_label.grid(row=3, column=0, columnspan=2)

root.mainloop()
```
